# Automated DCF Tool Demo
This notebook demonstrates how to use the `automated-dcf` library to value a public company.

```python
from automated_dcf import DCFModel
import matplotlib.pyplot as plt

# 1. Initialize the model
ticker = 'MSFT'
model = DCFModel(ticker)

# 2. Fetch data and run DCF
results = model.run_dcf(growth_rate=0.07)

print(f"Intrinsic Value for {ticker}: ${results['intrinsic_value']:.2f}")
print(f"Current Price: ${results['current_price']:.2f}")
print(f"Upside: {results['upside']*100:.2f}%")
```

```python
# 3. Plot results
model.plot_results()
```

```python
# 4. Export to Excel
model.export_to_excel(f"{ticker}_valuation.xlsx")
```
